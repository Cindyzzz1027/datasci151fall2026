```python
# Control statements
# if, elif, else
good_movies = [] 
ok_movies = []
bad_movies = []

cut_off_1 = 7.0 # minimum rating for "good movie"
cut_off_2 = 6.0


title_1 = "The Umbrellas of Cherbourg"
imdb_rating_1 = 7.8 

title_2 = "M3gan" 
imdb_rating_2 = 6.3 

title_3 = "Psycho" 
imdb_rating_3 = 8.5 

title_4 = "Melania"
imdb_rating_4 = 1.6

# Divide movies between the three lists
# print a warning if we encounter a bad movie

if imdb_rating_1 >= cut_off_1:
    good_movies = good_movies + [title_1] # add movie to list by concatenation 
elif imdb_rating_1 >= cut_off_2:
    ok_movies = ok_movies + [title_1]
else:
    bad_movies = bad_movies + [title_1]
    print("Warning! Bad movie alert:", title_1)

if imdb_rating_2 >= cut_off_1:
    good_movies = good_movies + [title_2] # add movie to list by concatenation 
elif imdb_rating_2 >= cut_off_2:
    ok_movies = ok_movies + [title_2]
else:
    bad_movies = bad_movies + [title_2]
    print("Warning! Bad movie alert:", title_2)


if imdb_rating_3 >= cut_off_1:
    good_movies = good_movies + [title_3] # add movie to list by concatenation 
elif imdb_rating_3 >= cut_off_2:
    ok_movies = ok_movies + [title_3]
else:
    bad_movies = bad_movies + [title_3]
    print("Warning! Bad movie alert:", title_3)


if imdb_rating_4 >= cut_off_1:
    good_movies = good_movies + [title_4] # add movie to list by concatenation 
elif imdb_rating_4 >= cut_off_2:
    ok_movies = ok_movies + [title_4]
else:
    bad_movies = bad_movies + [title_4]
    print("Warning! Bad movie alert:", title_4)


print(good_movies)
print(ok_movies)
print(bad_movies)
```

```python
# for item in items:

car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

for car_name in car_list:
    print("Dear customer, we are writing about your " + car_name + " vehicle.")
```

```python
# create an INDEX variable to keep track of where we are in the list
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

index = 1

for car_name in car_list:
    print("Dear " + car_name + " customer, your position is " + str(index) + " on the waitlist.")
    index = index + 1
```

```python
# use the "range" function
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

N = len(car_list) # number of elements

for index in range(N):
    print("Dear " + car_list[index] + " customer, your position is " + str(index + 1) + " on the waitlist.")
```

```python
# enumerate
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

for index, car_name in enumerate(car_list):
    print("Dear " + car_name + " customer, your position is " + str(index) + 
            " on the waitlist.")
```

```python
good_movies = []
cut_off = 7.0

movie_titles = ["The Umbrellas of Cherbourg", "M3gan", "Psycho"]
movie_ratings = [7.8, 6.3, 8.5]

thing_im_looping_over = range(len(movie_ratings))

for index in thing_im_looping_over:
    if movie_ratings[index] >= cut_off:
        good_movies = good_movies + [movie_titles[index]]

print(good_movies)
```
